```python
titanic = pd.read_csv("titanic.csv", sep="\t")

print(titanic.columns.tolist())
```

```python
titanic["Age"] = titanic["Age"].fillna(titanic["Age"].mean())

print("Missing Age values:", titanic["Age"].isnull().sum())
```

```python
titanic["Embarked"] = titanic["Embarked"].fillna(
    titanic["Embarked"].mode()[0]
)

print("Missing Embarked values:", titanic["Embarked"].isnull().sum())
```

```python
titanic["Cabin"] = titanic["Cabin"].fillna("Unknown")

print("Missing Cabin values:", titanic["Cabin"].isnull().sum())
```

```python
print(titanic.dtypes)
```

```python
titanic["Age"] = pd.to_numeric(
    titanic["Age"],
    errors="coerce"
)

print(titanic["Age"].dtype)
```

```python
print("Unique Survived values:")
print(titanic["Survived"].unique())

invalid_survived = titanic[
    ~titanic["Survived"].isin([0, 1])
]

print("\nInvalid Survived records:")
print(invalid_survived)
```

```python
print("Passenger class values:")
print(titanic["Pclass"].unique())
```
