Load Dataset

```python
titanic = pd.read_csv("titanic.csv")
print(titanic.head())
```

Fix col name

```python
titanic = pd.read_csv("titanic.csv", sep="\t")

print(titanic.columns.tolist())
```

Line plot

```python
plt.plot(titanic["PassengerId"], titanic["Fare"])

plt.title("Passenger Fare by Passenger ID")
plt.xlabel("Passenger ID")
plt.ylabel("Fare")

plt.show()
```

Scatter plot

```python
plt.scatter(titanic["Age"], titanic["Fare"])

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

Bar chart

```python
class_counts = titanic["Pclass"].value_counts().sort_index()

plt.bar(
    class_counts.index.astype(str),
    class_counts.values
)

plt.title("Number of Passengers by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

Histogram

```python
plt.hist(titanic["Age"].dropna(), bins=10)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

Pie chart

```python
survival_counts = titanic["Survived"].value_counts()

plt.pie(
    survival_counts.values,
    labels=["Didn't Survive", "Survived"],
    autopct="%1.0f%%"
)

plt.title("Passenger Survival Distribution")

plt.show()
```

Subplots

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Histogram
ax[0].hist(titanic["Age"].dropna(), bins=10)
ax[0].set_title("Age Distribution")
ax[0].set_xlabel("Age")
ax[0].set_ylabel("Frequency")

# Pie chart
survival_counts = titanic["Survived"].value_counts()

ax[1].pie(
    survival_counts.values,
    labels=["Did Not Survive", "Survived"],
    autopct="%1.1f%%"
)
ax[1].set_title("Survival Distribution")

plt.show()
```
